Data Normalization and Data Type Conversion

Part A : Data Normalization

```python
#Dataset
import pandas as pd
import numpy as np
data = {
    'Product': ['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
    'Price': [55000,20000,30000,2000,5000,45000],
    'Units_Sold': [25,60,35,80,50,15],
    'Discount':[10,15,12,20,10,8]
}
df = pd.DataFrame(data)
print(df)
```

```python
#1. Min-Max Normalization
#Min-Max Normalization for Price

df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())
df[['Product','Price','Price_MinMax']]

```

```python
#Normalize multiple columns
coils = ['Price','Units_Sold','Discount']
df_norm = (df[coils] - df[coils].min()) / (df[coils].max() - df[coils].min())
df_norm
```

```python
#2. Z-Score Normalization
df['Units_ZScore'] = (df['Units_Sold'] - df['Units_Sold'].mean()) / df['Units_Sold'].std()
df[['Product','Units_Sold','Units_ZScore']]
```

```python
#3. Decimal Scaling
df['Price_Decimal'] = df['Price'] / 100000
df[['Product','Price','Price_Decimal']]
```

Part B: Turning categoricalvariables into quantitative variables in python

```python
import pandas as pd
import numpy as np
data = {
'Order_ID':[101,102,103,104,105,106],
'Customer_Gender':['Male','Female','Female','Male','Female','Male'],
'Payment_Method':['UPI','Credit Card','Debit Card','COD','UPI','Debit Card'],
'Product_Category':['Electronics','Clothing','Home','Electronics','Beauty','Electronics'],
'City':['Pune','Mumbai','Delhi','Bangalore','Pune','Hyderabad'],
'Order_Value':[4500,2300,3200,5200,1800,4100]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Method 1: Label Encoding
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df['Gender_Label'] = label_encoder.fit_transform(df['Customer_Gender'])
df['City_Label'] = label_encoder.fit_transform(df['City'])
print(df)
```

```python
#Method 2: One Hot Encoding
df_encoded = pd.get_dummies(df, columns=['Payment_Method','Product_Category'])
print(df_encoded)
```

```python
#Method 3: Dummy Encoding (Drop First)
df_encoded = pd.get_dummies(df, columns=['Payment_Method'], drop_first=True)
print(df_encoded)
```
